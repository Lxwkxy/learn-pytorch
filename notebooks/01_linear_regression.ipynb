# Linear Regression with PyTorch

This notebook trains a model to learn the relationship y = 2x + 1 from example data.

```python
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
```

## 1. Prepare data

TensorDataset pairs each input x with its target y, while DataLoader splits the data into batches.

```python
x_train = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y_train = torch.tensor([[3.0], [5.0], [7.0], [9.0]])

train_dataset = TensorDataset(x_train, y_train)
train_dataloader = DataLoader(train_dataset, batch_size=2, shuffle=True)
```

## 2. Build the model

nn.Linear(1, 1) learns the equation y = wx + b.

```python
model = nn.Linear(1, 1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05)
```

## 3. Train

Each iteration follows the same cycle: predict, calculate loss, run backpropagation, and update the weights.

```python
epochs = 100

for epoch in range(epochs):
    for x_batch, y_batch in train_dataloader:
        prediction = model(x_batch)
        loss = loss_fn(prediction, y_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    if (epoch + 1) % 25 == 0:
        print(f'Epoch {epoch + 1}: loss = {loss.item():.6f}')
```

## 4. Inference

```python
model.eval()
x_test = torch.tensor([[5.0]], dtype=torch.float32)

with torch.no_grad():
    test_prediction = model(x_test)

print(f'Prediction for x = 5: {test_prediction.item():.4f}')
```

## 5. Save and load model weights

```python
model_path = 'linear_model.pth'
torch.save(model.state_dict(), model_path)

loaded_model = nn.Linear(1, 1)
loaded_model.load_state_dict(torch.load(model_path, weights_only=True))
loaded_model.eval()

with torch.no_grad():
    loaded_prediction = loaded_model(x_test)

print(f'Loaded model prediction: {loaded_prediction.item():.4f}')
```

## Result

The model learns the relationship y = 2x + 1 and predicts a value close to 11 for x = 5.
